# **New Section**

```python
import tensorflow_hub as hub
import tensorflow as tf
from matplotlib import pyplot as pp
import numpy as np
import cv2
```

```python
model = hub.load('https://tfhub.dev/google/magenta/arbitrary-image-stylization-v1-256/2')
```

```python
def load_image(img_path):
    img = tf.io.read_file(img_path)
    img = tf.image.decode_image(img, channels=3)
    img = tf.image.convert_image_dtype(img, tf.float32)
    img = img[tf.newaxis, :]
    return img
```

# **Load images**

```python
content_image = load_image('/content/IMG_20241104_183038_172.jpg')
style_image = load_image('/content/download.jfif')
```

# **Plot the graph of the 1st image**

```python
pp.imshow(np.squeeze(content_image))
pp.show()
```

# **Plot the grapgh of the 2nd image**

```python
pp.imshow(np.squeeze(style_image))
pp.show()
```

# **Style transfer**

```python
stylized_image = model(tf.constant(content_image), tf.constant(style_image))[0]
```

```python
pp.imshow(np.squeeze(stylized_image))
pp.show()
```

# **Save the generated image**

```python
cv2.imwrite('generated_image.jpg', cv2.cvtColor(np.squeeze(stylized_image)*255, cv2.COLOR_BGR2RGB))
```
