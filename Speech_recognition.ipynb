```python
import speech_recognition as sr
```

```python
from pydub import AudioSegment
```

```python
import os
```

```python
# Function to transcribe an audio file
def transcribe_audio_file(audio_file_path):
    r = sr.Recognizer()
    with sr.AudioFile(audio_file_path) as source:
        audio = r.record(source)
    try:
        text = r.recognize_google(audio)
        return text
    except sr.UnknownValueError:
        return "Could not understand audio"
    except sr.RequestError as e:
        return f"Could not request results; {e}"
```

```python
# Function to split audio into segments
def split_audio(audio_file_path, segment_duration=60):
    audio = AudioSegment.from_file(audio_file_path)
    total_duration = len(audio)
    segments = []
    for i in range(0, total_duration, segment_duration * 1000):
        segment = audio[i:i + segment_duration * 1000]
        segment_path = f"segment_{i // (segment_duration * 1000)}.wav"
        segment.export(segment_path, format="wav")
        segments.append(segment_path)
    return segments
```

```python
# Main execution
audio_file = "/content/quothello-therequot-158832.wav"
```

```python
segments = split_audio(audio_file, segment_duration=60)
```

```python
all_text = ""
for segment_path in segments:
    text = transcribe_audio_file(segment_path)
    all_text += text + " "
    print(f"Transcribed segment: {segment_path}")
```
